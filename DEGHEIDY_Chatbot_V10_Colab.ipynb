# DEGHEIDY Chatbot V10 — Fresh Colab copy

Open in **Google Colab** and run the code cell. If GPU is unavailable, it starts in CPU voice-only mode. The first run installs dependencies.

```python
# DEGHEIDY V10 — GPU chat/large-v3 speech or CPU voice-only/small speech.
import torch, subprocess, sys, io, base64, os, zipfile
is_gpu = bool(torch.cuda.is_available())
print('GPU available:', is_gpu)
need = ['gradio>=5.0,<7.0','faster-whisper>=1.2.0,<2.0',
        'webrtcvad-wheels>=2.0.13','librosa>=0.10.1','numpy>=1.24']
if is_gpu:
    need += ['transformers>=4.57,<5.0','accelerate>=1.0','bitsandbytes>=0.45']
print('Installing dependencies...')
subprocess.check_call([sys.executable, '-m','pip','install','-q', *need])
payload = ''
with zipfile.ZipFile(io.BytesIO(base64.b64decode(payload))) as archive:
    archive.extractall('/content')
sys.path.insert(0,'/content/DEGHEIDY_Chatbot_V10_SpeechFix')
from app import load_chat_model, make_ui
if is_gpu:
    print('GPU mode: loading Qwen3 4B; high-quality Whisper large-v3 loads on first voice use.')
    load_chat_model()
else:
    print('CPU mode: speech recognition only (Whisper small); AI Send is disabled.')
print('Open the HTTPS Gradio share URL in a separate Chrome tab for microphone permissions.')
make_ui().queue(default_concurrency_limit=1).launch(share=True, inline=True, debug=True, show_error=True)
```
